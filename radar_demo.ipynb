```python
import matplotlib
matplotlib.use("TkAgg")

import numpy as np
import matplotlib.pyplot as plt
import time

from radar.radar_3d import Radar3D
from radar.radar_simulator import RadarSimulator
from radar.radar_environment import OrbitalTarget, RadarEnvironment


# ============================================================
# SIMULACIÓN 3D EN TIEMPO REAL
# ============================================================

CONFIG = {
    "trail_seconds": 3,

}


radar = Radar3D(
    position=[0, 0, 0],         # posición del radar
    az0_deg=30,                 # dirección inicial del haz
    el0_deg=10,                 # elevación inicial del haz

    bw_az_deg=6,                # ancho del haz en azimut
    bw_el_deg=4,                # ancho del haz en elevación

    R=2000e3,                   # alcance máximo (2000 km)
    rpm=12,                     # revoluciones por minuto
    el_min_deg=10,              # elevación mínima del barrido
    el_max_deg=75,              # elevación máxima del barrido
)


# Agregar un objetivo orbital para probar la detección
env = RadarEnvironment(radar)
sat = OrbitalTarget(
    a=500e3,                    # semieje mayor
    e=0.001,                    # excentricidad
    i=np.radians(45),           # inclinación
    raan=np.radians(10),        # nodo ascendente
    argp=np.radians(0),         # argumento del periapsis
    nu0=np.radians(0)           # anomalía verdadera inicial
)

env.add_target(sat)


# Crear el simulador con el radar y la configuración de la estela
sim = RadarSimulator(radar, env, trail_seconds=CONFIG["trail_seconds"])
sim.run()

```
